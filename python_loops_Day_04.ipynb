```python
#Understanding range()
for i in range(5):
  print(i)

for j in range(1,6):
  print(j)

for k in range(2,11,2):
  print(k)
```

```python
#Print multiplication table of 5
for i in range(1,11):
  print(5,"x",i ,"=", 5*i)
```

```python
#Using a for loop with a list
fruits = ["apple" ,"banana","cherry"]
for x in fruits:
  print(x)
```

```python
#Using a for loop with a string
name = "shivani"
for i in name:
  print(i)
```

```python
#What is a while loop?
#A while loop repeats a block of code as long as a condition is true.
#Always make sure your while loop has a way to eventually stop. Otherwise, it can become an infinite loop.
count = 1

while count <= 5:
  print(count)
  count = count+1
```

```python
#Multiplication table of 7
num = 1

while num <= 10:
  print(7, "x", num, "=", 7*num)
  num = num+1
```

```python
# The break statement
# break immediately stops the loop.

for i in range(1,11):
  if i == 5:
    break
  print(i)

# The continue statement
# continue skips the current iteration and moves to the next one.

for j in range(12, 22):
  if j == 16:
    continue
  print(j)
```

```python
#Use a for loop to print numbers from 1 to 20

for i in range(1,21):
  print(i)
```

```python
#Ask the user to enter a number and print its multiplication table from 1 to 10.

number = int(input("enter any number for multiplication table"))

for i in range(1,11):
  print(number,"x",i,"=",number*i)
```

```python
#Use a for loop to calculate the sum of numbers from 1 to 100

total = 0
for i in range(1,101):
  total = total+i

print(total)
```

```python
#Use a for loop to print all even numbers between 1 and 50.

for i in range(1,51,2):
  print(i+1)
```

```python
#Use a while loop to print numbers from 10 down to 1.
count = 10
while count >= 1:
  print(count)
  count = count-1
```

```python
#Ask the user to enter a positive integer and calculate its factorial using a for loop.

number = int(input("enter any positive integer:"))

if number < 0:
  print("enter positive integer")
else:
  factorial = 1
  for i in range(1,number+1):
    factorial = factorial*i
  print("the factorial of given number is :",factorial)
```

```python
#Print numbers from 1 to 10, but stop the loop when the number reaches 6.
for i in range(1,11):
  if i == 6:
    break
  print(i)


```

```python
#Print numbers from 1 to 10, but skip number 5.
for k in range(1,11):
  if k == 5:
    continue
  print(k)
```

```python
#Write a program with a fixed secret number of 7. Ask the user to enter a number repeatedly using a while loop until they guess correctly. Print "Correct guess!" when they succeed.
secret_number = 7
guess = 0

while guess != secret_number:
    guess = int(input("Guess the secret number: "))

print("Correct guess!")
```
