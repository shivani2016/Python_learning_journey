```python
marks = [75, 80, 65, 90, 85]
#Find the total marks using sum().
total_marks = sum(marks)
print("Total marks are", total_marks)


#Find the highest marks using max().
highest_marks = max(marks)
print("highest marks are", highest_marks)

#Find the lowest marks using min().
lowest_marks = min(marks)
print("lowest_marks are", lowest_marks)

#Find the number of subjects using len()
number_of_subjects = len(marks)
print("number of subjects are", number_of_subjects)

#Calculate the average marks.
average_marks = total_marks / number_of_subjects
print("average marks are", average_marks)
```

```python
import random

fruits = ["apple","banana","cherry","kiwi","pear"]        # Create a list of five fruits
print(random.choice(fruits))                              # 1. Select one random fruit

random_number = random.randint(1,10)                      # 2. Generate a random number between 1 and 10 (inclusive)
print(random_number)

random.shuffle(fruits)                                   # 3. Shuffle the fruit list in place
print(fruits)
```

```python
from datetime import date, datetime

# 1. Display today's date
today = date.today()
print("1. Today's date:", today)

# 2. Display the current date and time
now = datetime.now()
print("2. Current date and time:", now)

# 3. Display the current year, month, and day
print("3. Year:", today.year, "| Month:", today.month, "| Day:", today.day)

# 4. Display the date in DD-MM-YYYY format
formatted_date = today.strftime("%d-%m-%Y")
print("4. DD-MM-YYYY format:", formatted_date)
```
