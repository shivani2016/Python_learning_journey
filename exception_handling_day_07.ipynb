Exception-handling :Meaningful message instead of an unhandled crash--
try – contains code that might produce an exception.
except – contains code that handles the exception.

```python
try:
  a = 10
  b = 0
  result = a/b
  print("Result:",result)
except ZeroDivisionError:
  print("Error,Division by zero is not allowed")
```

```python
try:
  a = int(input("enter a number :"))
  print("entered number is:",a)
except ValueError:
  print("invalid input")
```

Handling multiple exceptions--We can handle each error separately using multiple except blocks.

```python
try:
  a = int(input("enter first number :"))
  b = int(input("enter second number :"))       #Important: Python executes the first matching except block and skips the remaining ones.
  result = a/b
  print("result is :",result)
except ValueError:
  print("invalid input")
except ZeroDivisionError:
  print("cannot be divided by zero")
```

```python
#The else block executes only when no exception occurs inside the try block.

try:
  number = int(input("enter any number:"))                  #try – attempts to execute code.
except ValueError:                                          #except – handles an exception.
  print("invalid input")                                    #else – executes when no exception occurs.
else:                                                       #The finally block executes whether an exception occurs or not.
  print("the number u have entered is :", number)  #Sometimes we want to create an exception ourselves when a particular condition is not satisfied.
  print("the program executed successfully")
finally:
  print("thank you for using google collab for coding")
```

```python
age = 22

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")

    print("You are eligible")

except ValueError as e:
    print("Error:", e)
```

```python
def withdraw(balance, amount):

  if amount < 0:
    raise ValueError("amount cannot be negative and it must be positive")  #here v deliberately raised an exception
  if amount > balance:
    raise ValueError("insufficient balance")

  return balance - amount

print(withdraw(10000,2000))
```

```python
#TypeError
try:
  result = "python"+10
except TypeError:
  print("cannot add string with integer")

#IndexError
try:
    fruits = ["Apple", "Mango", "Banana"]
    print(fruits[5])

except IndexError:
    print("List index does not exist")
```

```python
# Assignment 5 – Create a function called check_age(age)
# Requirements: If age is below 18, raise ValueError. Otherwise, return "Eligible".

def check_age(age):
  if age < 18:
    raise ValueError("The age must be 18 or above")
  return "Eligible"

try:
  age = int(input("enter your age :"))
  result = check_age(age)
  print(result)
except ValueError as e:
  print("Error",e)
```

```python
# Assignment 6 – Handle IndexError and ValueError
# Requirements: Create a list, ask for an index, display the corresponding element, and handle errors.

numbers = [10,20,30,40,50]

try:
  index = int(input("enter an index to display corresponding element(0 to 4): "))
  print("selected element :",numbers[index])

except ValueError:
  print("invalid input ........please enter integer index")
except IndexError:
  print("index is out of range")
```

```python
# Assignment 7 – Bank withdrawal program
# Job-oriented challenge
# Requirements: Initial balance ₹10,000, validate withdrawal amount, handle exceptions, display remaining balance, and use finally.

initial_balance = 10000
try:
  amount = float(input("enter amount to withdraw: "))
  if amount < 0:
    raise ValueError("withdrawal amount should be more than zero (positive)")
  if amount > initial_balance:
    raise Exception("insufficient balance")
  final_balance = initial_balance - amount
  print("withdrawal successful")
  print("withdrawal amount : ",amount)
  print("remaining balance is: ",final_balance)
except ValueError as e:
    print("Invalid withdrawal:", e)
except Exception as e:
    print("Transaction failed:", e)
finally:
    print("Thank you for banking with us.")
```
