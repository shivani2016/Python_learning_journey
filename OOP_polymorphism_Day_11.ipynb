```python
# All three classes have the same method name:
# But each class performs a different action.
# This is polymorphism!
class Dog:
  def speak(self):
    print("Dog barks")
class Cat:
  def speak(self):
    print("Cat meows")
class Cow:
  def speak(self):
    print("Cow makes sounds ")

dog1 = Dog()
cat1 = Cat()
cow1 = Cow()

dog1.speak()
cat1.speak()
cow1.speak()
```

method overriding --> When a child class provides its own implementation of a method that is already defined in its parent class, it is called method overriding.

```python
class Animal:                                    #Animal has a method called speak().
  def speak(self):                               #Dog inherits from Animal but overrides speak().
    print("Animal makes sound")                  #Cat also inherits from Animal but overrides speak()
class Dog(Animal):                               #When we call speak(), Python executes the method belonging to the actual object's class.
  def speak(self):
    print("Dog barks")
class cat(Animal):
  def speak(self):
    print("Cow makes sounds ")

animal1 = Animal()
dog1 = Dog()
cat1 = Cat()

animal1.speak()
dog1.speak()
cat1.speak()
```

Polymorphism using a common function-->We can create one function that works with different objects.

```python
class Developer:
  def work(self):
    print("He writes code")

class Designer:
  def work(self):
    print("He draws designs")

class Tester:
  def work(self):
    print("he tests applications")

def employee_work(employee):  #in method -->the first parameter should always be--> self
  employee.work()             #method is called by --> obj.method(arg)

dev1 = Developer()
des1 = Designer()
tes1 = Tester()

employee_work(dev1)
employee_work(des1)
employee_work(tes1)
```

```python
# Assignment 2 – Shape Area
# 1. Create a parent class Shape.
# 2. Add a method called area().
# 3. Create child classes Rectangle and Circle.
# 4. Override the area() method in both child classes.
# 5. Rectangle should calculate length × breadth.
# 6. Circle should calculate 3.14 × radius × radius.
# 7. Create objects and display both areas

class Shape:
  def area(self):
    print("calculating area")

class Rectangle(Shape):
  def __init__(self,length, breadth):
    self.length = length
    self.breadth = breadth
  def area(self):
    return self.length * self.breadth
class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius
  def area(self):
    return 3.14 * self.radius * self.radius

r1 = Rectangle(10,5)
c1 = Circle(7)

print("Area of rectangle : ",r1.area())
print("Area of circle : ", c1.area())
```

```python
# Assignment 5 – Calculator
# 1. Create a class called Calculator.
# 2. Add methods for addition, subtraction, multiplication and division.
# 3. Create an object.
# 4. Call all four methods with different numbers.
# 5. Display the results.

class Calculator:
  def addition(self,a,b):
    return a+b
  def subtraction(self,a,b):
    return a-b
  def multiplication(self,a,b):
    return a*b
  def division(self,a,b):
    if b == 0:
      return "cannot be divided by 0"
    return a/b
calc1 = Calculator()

print("Addition:", calc1.addition(20, 10))

print("Subtraction:", calc1.subtraction(20, 10))

print("Multiplication:", calc1.multiplication(20, 10))

print("Division:", calc1.division(20, 10))
```

```python
class CreditCardPayment:
  def make_payment(self):
    print("Payment made through Credit Card")

class UPIPayment:
  def make_payment(self):
    print("Payment made through UPI")

class NetBankingPayment:
  def make_payment(self):
    print("Payment made through Net Banking")

def process_payment(payment):
  payment.make_payment()

c1 = CreditCardPayment()
u1 = UPIPayment()
n1 = NetBankingPayment()

process_payment(c1)
process_payment(u1)
process_payment(n1)
```
