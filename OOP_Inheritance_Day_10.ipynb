```python
class Parent:                             #Parent is the parent class.
  def display(self):
    print("this is the parent class :")

class Child(Parent):             #means Child inherits from Parent.
  def show(self):
    print("this is child class :")

obj = Child()
obj.display()                 #works even though we did not define display() inside Child.
obj.show()
```

```python
#Real-world example – Animal
#Let's create a parent class called Animal and a child class called Dog.

class Animal:
  def eat(self):
    print("All animals can eat")
class Dog(Animal):
  def bark(self):              #the Dog class has its own method bark() and inherits the eat() method from Animal.
    print("All Dogs can bark")
obj1 = Dog()
obj1.eat()
obj1.bark()
```

```python
class Animal:
  def __init__(self,name):
    self.name = name
  def display(self):
    print("Animal name :", self.name)

class Dog(Animal):
  def bark(self):
    print(self.name, "is barking")  #Dog object uses the constructor of Animal because Dog does not have its own constructor.
dog1 = Dog("Tommy")
dog1.display()
dog1.bark()
```

Using super() in inheritance
What if both parent and child classes have their own constructors?
We use super() to call the parent class constructor.

```python
class Animal:
  def __init__(self,name):
    self.name = name

class Dog(Animal):
  def __init__(self,name,breed):
    super().__init__(name)     #super().__init__(name) calls the parent class constructor.
    self.breed = breed         #self.name is initialized in the parent class.
  def display(self):           #self.breed is initialized in the child class.
    print("Name : ", self.name)
    print("Breed : ", self.breed)
dog1 = Dog("Tommy","labrador")
dog1.display()
```

```python
# Assignment 1 – Person and Student
# 1. Create a parent class Person.
# 2. Add name and age using __init__().
# 3. Create a child class Student.
# 4. Add a student ID in the child class.
# 5. Create an object and display all three details.

class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
class Student(Person):
  def __init__(self,name,age,student_id):
    super().__init__(name,age)
    self.student_id = student_id
  def display(self):
    print("student name : ",self.name)
    print("student age : ", self.age)
    print("student id : ",self.student_id)
s1 = Student("shivani",25,101)
s1.display()
```

```python
# Assignment 2 – Employee and Manager
# 1. Create a parent class Employee.
# 2. Add employee name and salary.
# 3. Create a child class Manager.
# 4. Add department in the child class.
# 5. Display employee name, salary and department.

class Employee:
  def __init__(self,name,salary):
    self.name = name
    self.salary = salary
class Manager(Employee):
  def __init__(self,name,salary,department):
    super().__init__(name,salary)
    self.department = department
  def display(self):
    print("employee name : ",self.name)
    print("salary : ", self.salary)
    print("department : ", self.department)
m1 = Manager("shivani",50000, "IT")
m1.display()
```

```python
# Assignment 3 – Vehicle and Car
# 1. Create a parent class Vehicle.
# 2. Add brand and speed.
# 3. Create a child class Car.
# 4. Add fuel type.
# 5. Use super() to initialize the parent properties.
# 6. Display all details.

class Vehicle:
  def __init__(self,brand,speed):
    self.brand = brand
    self.speed = speed

class Car(Vehicle):
  def __init__(self,brand,speed,fuel_type):
    super().__init__(brand,speed)
    self.fuel_type = fuel_type
  def display_details(self):
    print("The car brand : ",self.brand)
    print("The car speed : ",self.speed ,"km/hr")
    print("The car fuel type : ",self.fuel_type)
v1 = Car("toyota",100,"Petrol")
v1.display_details()
```

```python
# Assignment 4 – Animal, Dog and Cat
# 1. Create a parent class Animal.
# 2. Add a method eat().
# 3. Create child classes Dog and Cat.
# 4. Add their own methods bark() and meow().
# 5. Create objects of both child classes.
# 6. Call all their methods.

class Animal:
  def eat(self):
    print("Animals can eat")
class Dog(Animal):
  def bark(self):
    print("All dogs will bark")
class Cat(Animal):
  def meow(self):
    print("All cats will meow")

d1 = Dog()
c1 = Cat()

print("Dog details :")
d1.eat()
d1.bark()

print("\n Cat details :")
c1.eat()
c1.meow()
```

```python
# Assignment 5 – Bank Account and Savings Account
# 1. Create a parent class BankAccount.
# 2. Add account holder name and balance.
# 3. Add a method to display account details.
# 4. Create a child class SavingsAccount.
# 5. Add an interest rate.
# 6. Use super() to initialize parent properties.
# 7. Add a method to calculate annual interest.

class BankAccount:
  def __init__(self, holder_name, balance):
    self.holder_name = holder_name
    self.balance = balance

  def display_account_details(self):
    print("The account holder name : ",self.holder_name)
    print("The bank balance is : ", self.balance)

class SavingsAccount(BankAccount):
  def __init__(self, holder_name, balance, interest_rate):
    super().__init__(holder_name, balance)
    self.interest_rate = interest_rate

  def calculate_annual_interest(self):
    interest = self.balance * self.interest_rate /100
    return interest

  def display_details(self):
    self.display_account_details()
    print("Interest Rate:", self.interest_rate, "%")
    print("Annual Interest:", self.calculate_annual_interest())

account1 = SavingsAccount("Anjali", 50000, 5)

account1.display_details()
```

```python
# Bonus Assignment – Employee Management System
# Create a parent class Employee and two child classes:
# - Developer – add programming language.
# - Designer – add design tool.
# Use inheritance and super() to display the details of both employees.

class Employee:
  def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

  def display(self):
        print("Employee Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Salary:", self.salary)
class Developer(Employee):
  def __init__(self, name, employee_id, salary, language):
        super().__init__(name, employee_id, salary)
        self.language = language

  def display_details(self):
        print("Developer Details:")
        self.display()
        print("Programming Language:", self.language)
class Designer(Employee):
   def __init__(self, name, employee_id, salary, tool):
        super().__init__(name, employee_id, salary)
        self.tool = tool

   def display_details(self):
        print("Designer Details:")
        self.display()
        print("Design Tool:", self.tool)
developer1 = Developer("Arun", "DEV101", 60000, "Python")
designer1 = Designer("Meena", "DES101", 45000, "Figma")
developer1.display_details()
print()
designer1.display_details()
```
