3. Python Sets----unique values,No duplicates

```python
numbers = {10,10,10,20,20,30}     #no duplicates, curly brackets , mutable
print(numbers)
numbers.add(40)
print(numbers)
numbers.remove(40)
print(numbers)
```

```python
names = {1,2,3,4,5,5,4,3,2,1}           #removes the duplicates
unique_names = set(names)
print(unique_names)
```

```python
student1 = {"Python", "Java", "SQL"}     #Union: Combines all unique elements.
student2 = {"Python", "C++", "SQL"}      #intersection : finds common elements
                                        #Difference: Finds elements present in the first set but not the second.
print(student1.union(student2))
print(student1.intersection(student2))
print(student1.difference(student2))
```

4. Python Dictionaries -------- key value pairs

```python
student = {
    "name": "shivani",
    "age" : 20,
    "city" : "pune",
    "course" : "python"
}                                 #curly brackets,mutable, key should be unique, values can be duplicated ,
student["Grade"] = "A"            #add new key-value pair
print(student)
print(student["name"])
print(student["age"])             #Tuple Multiple values that you cannot change.

                                   #Set Unique values without fixed ordering.
print(student["city"])
print(student["course"])
print(student["Grade"])

student["course"] = "Javascript"    #updating an existing value
print(student)
```

```python
student.pop("city")                    #remove key-value pair
print(student)
```

```python
print(student.get("name"))     #get()---	Safely retrieves a value
print(student.get("salary","not available"))   #get() avoids a KeyError when a key is missing.
```
