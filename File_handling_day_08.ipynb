File handling ---> working with files using python

```python
file = open("student.txt","w")         #Writing into a file – "w" mode

file.write("Hello everyone\n")
file.write("My name is shivani\n")
file.write("i am learning python\n")

file.close()
print("created file successfully")
```

```python
file = open("student.txt","r")   #Reading a file – "r" mode
content = file.read()
print(content)
file.close()
```

```python
file = open("student.txt","r")
print(file.readline())         #reads one line

file.close()
```

```python
# Appending information – "a" mode
# Suppose you want to add another line without deleting the existing information.

file = open("student.txt","a")
file.write("\n practice python everyday")
file.close()
print("new information added")
```

```python
file = open("student.txt","r")
print(file.read())         #"w" – Replaces existing content.
file.close()               #"a" – Adds content at the end.          #"r" – Reads existing content.
```

```python
#The better way – with open()
#Because Python automatically closes the file when the with block finishes. You don't need to write close() manually.
with open("student.txt","r") as file:
  print(file.read())                    #Learn to use with open() because it is the recommended approach for everyday file handling.
```

Handling file errors

```python
try:
  with open("unknown.txt","r") as file:
    print(file.read())
                                    #Python raises FileNotFoundError.---- if file is not found
except FileNotFoundError:
  print("File not found")
```

Practical example – Student Information System

```python
#taking student details
name = input("Enter your name : ")
age = int(input("Enter your age : "))
course = input("Enter your course : ")

#saving information into a file
with open("student_info.txt","w") as file:
  file.write("student information\n")
  file.write(f"Name : {name}\n")
  file.write(f"Age : {age}\n")
  file.write(f"Course : {course}\n")
print("student details saved successfully")

# Reading information from the file
with open("student_info.txt","r") as file:
  print("\n saved student details are :")
  print(file.read())
```

```python
#Assignment 2 – Shopping List using Append Mode

# Writing five shopping items
with open("shopping_list.txt", "w") as file:
  file.write("Rice\n")
  file.write("Milk\n")
  file.write("Fruits\n")
  file.write("sugar\n")
  file.write("eggs\n")
print("shopping list created successfully")

# Appending two additional items
with open("shopping_list.txt", "a") as file:
  file.write("butter\n")
  file.write("bread\n")
print("Two new items are added to the shopping list ")

# Reading all seven items
with open("shopping_list.txt", "r") as file:
  print("\nMy shopping list is :")
  print(file.read())
```
