```python
class Student:

  def __init__(self,name,age):                       #variables
    self.name = name
    self.age = age

  def display_details(self):                        #method
    print("Student Name:", self.name)
    print("Student Age:", self.age)

student1 = Student("Shivani", 25)                 #creating object
student2 = Student("Mahesh", 35)
student3 = Student("Priya", 20)

student1.display_details()                         #calling method
print()
student2.display_details()
print()
student3.display_details()
print()
```

```python
class Employee:

  def __init__(self,name,salary):                       #variables--# ఇవి Instance Variables (ప్రతి ఆబ్జెక్ట్ కి వేరుగా ఉంటాయి)
    self.name = name
    self.salary = salary

  def display_details(self):                            #methods--• ఆబ్జెక్ట్ ఏం చేస్తుంది? (ఉదా: వివరాలు చూపించడం)
    print("Employee name :", self.name)
    print("Employee salary :", self.salary)

  def calculate_bonus(self):
    bonus = self.salary*0.1
    print("Employee bonus :", bonus)

emp1 = Employee("Shivani",50000)                     #creating object----# రెండు వేర్వేరు ఆబ్జెక్ట్స్ (Students) తయారు చేద్దాం
emp2 = Employee("Mahesh", 60000)

emp1.display_details()                               #calling methods

emp1.calculate_bonus()
print()
emp2.display_details()

emp2.calculate_bonus()


```

```python
#Creating and Updating Object Attributes

class Employee:                                            #instance variables
  def __init__(self,name,salary):
    self.name = name
    self.salary = salary

  def display_details(self):                                 #instance methods
    print("Employee name: ", self.name)
    print("Employee salary: ", self.salary)

employee1 = Employee("Shivani",60000)                      #creating object
print("before updating:")
employee1.display_details()                                   #calling methods

#updating salary
employee1.salary = 70000
print("after updating:")
employee1.display_details()
```

```python
#Real-world example – Bank Account
# Let's build a small bank account program using OOP.

class BankAccount:                                        #Created a class called BankAccount.

  def __init__(self,account_holder,balance):             #instance variables
    self.account_holder = account_holder
    self.balance = balance
  def deposit(self,deposit_amount):                        #Created a deposit method.
    self.balance = self.balance + deposit_amount
    print("Amount deposited successfully")
  def withdraw(self,withdraw_amount):                      #Created a withdrawal method.
    if withdraw_amount <= self.balance:
      self.balance = self.balance - withdraw_amount
      print("amount withdrawn successfully")
    else:
      print("Insufficient balance")
  def display_balance(self):                                #Created a method to display the balance.
    print("Account holder:", self.account_holder)
    print("Account balance:", self.balance)

account1 = BankAccount("shivani",10000)                      #Created an object and performed operations.

account1.display_balance()

account1.deposit(5000)

account1.withdraw(2000)

print("\nfinal account Details is ")
account1.display_balance()
```

```python
# Assignment 3 – Salary Calculator
# Create a class called Employee.
# - Accept employee name and salary.
# - Create a method to calculate a 10% bonus.
# - Create another method to calculate total salary (salary + bonus).
# - Display employee name, original salary, bonus and total salary.

class Employee():

  def __init__(self,name,salary):
    self.name = name
    self.salary = salary
  def calculate_bonus(self):
    bonus = self.salary*0.10
    return bonus
  def calculate_total_salary(self):
    bonus = self.calculate_bonus()
    total_salary = self.salary + bonus
    return total_salary
  def display(self):
    print("Employee name :", self.name)
    print("Employee Salary :",self.salary)
    print("Bonus :",self.calculate_bonus())
    print("total salary is :",self.calculate_total_salary())
# Taking user input

name = input("Enter employee name: ")
salary = float(input("Enter employee salary: "))


# Creating object

emp1 = Employee(name, salary)


# Displaying details

print("\nEmployee Salary Details:")
emp1.display()
```

```python
# Assignment 4 – Bank Account
# Create a class called BankAccount.
# - Initialize account holder and balance.
# - Create a deposit() method.
# - Create a withdraw() method.
# - Create a display_balance() method.
# - Handle insufficient balance.
# - Test your program with multiple deposits and withdrawals.

class BankAccount():

  def __init__(self,account_holder,balance):
    self.account_holder = account_holder
    self.balance = balance
  def deposit(self,deposit_amount):
    if deposit_amount > 0:
      self.balance = self.balance + deposit_amount
      print("amount deposited successfully")
    else:
      print("please enter valid amount to deposit")
  def withdraw(self,withdraw_amount):
    if withdraw_amount < 0:
      print("please enter valid amount to withdraw")
    elif withdraw_amount < self.balance:
      self.balance = self.balance - withdraw_amount
      print("amount withdrawn successfully")
    else:
      print("Insufficient balance ")
  def display_balance(self):
    print("Account holder :",self.account_holder)
    print("Available balance :",self.balance)

account1 = BankAccount("Shivani", 80000)            # Creating bank account
#account2 = BankAccount("Mahesh", 50000)

print("Initial account details :")                    # Initial balance
account1.display_balance()

print("\ndepositing money:")                       # Depositing money
account1.deposit(2000)

print("\nwithdrawn amount: ")                      # Withdrawing money
account1.withdraw(4000)

print("\nTrying Another Withdrawal:")                # Trying to withdraw more than available balance
account1.withdraw(90000)

print("\nFinal Account Details:")
account1.display_balance()
```

```python
# Assignment 5 – Product Management
# Create a class called Product.
# - Initialize product name, price and quantity.
# - Create a method to calculate total product value (price × quantity).
# - Create a method to update quantity.
# - Display product details and total value.
# - Create at least three product objects.

class Product:

  def __init__(self,name, price, quantity):
    self.name = name
    self.price = price
    self.quantity = quantity
  def calculate_total_product_value(self):
    return self.price*self.quantity
  def update_quantity(self,new_quantity):
    self.quantity = new_quantity
  def display_product_details(self):
    print("Product name :",self.name)
    print("Price:", self.price)
    print("Quantity:", self.quantity)
    print("Total Value:", self.calculate_total_product_value())

# Creating three product objects

product1 = Product("Laptop", 50000, 2)
product2 = Product("Mobile", 20000, 3)
product3 = Product("Headphones", 2000, 5)

# Displaying product details

print("Product 1 Details:")
product1.display_product_details()

print("\nProduct 2 Details:")
product2.display_product_details()

print("\nProduct 3 Details:")
product3.display_product_details()
# Updating product quantity

print("\nUpdating Laptop Quantity:")
product1.update_quantity(4)

print("\nUpdated Laptop Details:")
product1.display_product_details()

```
